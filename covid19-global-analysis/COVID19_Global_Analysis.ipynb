**3.3 Mortality Rate.**

Mortality rate is the measure of the number of deaths to the total population.

```python
#Creating a column called mortality_rate
summary_df["mortality_rate"] = summary_df["total_deaths"] / summary_df["population"] * 100
summary_df.head()
```

**3.4 Fatality Rate.**

Fatality Rate is the measure of deaths caused by a certain disease or thing, compared to the tottal number of people affected by the disease or thing

```python
summary_df["fatality_rate"] = summary_df["total_deaths"] / summary_df["total_confirmed"] * 100

summary_df.head()
```

**5.1 Coronavirus Confirmed Cases by Countries.**

```python
def choropleth(Stats, ColorbarTitle, GraphTitle):
    fig = go.Figure(
        data = go.Choropleth(
            locations = summary_df['country'],
            locationmode = 'country names',
            z = Stats,
            text = summary_df['country'],
            colorscale = 'reds',
            autocolorscale = False,
            reversescale = False,
            marker_line_color = 'darkgray',
            marker_line_width = 0.5,
            colorbar_title = ColorbarTitle,
        )
    )

    fig.update_layout(
        title_text = GraphTitle,
        geo = dict(
            showframe = False,
            showcoastlines = True,
            projection_type = 'equirectangular'
        ),
    )

    fig.show()
```

```python
def pie_chart(continent, text):
    trace = go.Pie(
        labels = [
            'Total Confirmed',
            'Total Deaths',
            'Total Recovered',
            'Total Active',
            'Critical Cases'
        ],
        values = [
            continent.total_confirmed.sum(),
            continent.total_deaths.sum(),
            continent.total_recovered.sum(),
            continent.active_cases.sum(),
            continent.serious_or_critical.sum()
        ],
        title_font_size = 20,
        hovertemplate = "<b>%{label}</b><b><br>%{value}<br><i>%{percent}</i>",
        textinfo = 'percent',
        textposition = 'inside',
        showlegend = True,
        marker = dict(
            colors = [
                "#ff5148",
                "#000201",
                "#61f7ff",
                "#ffb861",
                "#61ffb8"
            ],
            line = dict(
                color = '#000000',
                width = 2
            )
        )
    )
    fig = go.Figure(data=[trace])
    fig.update_layout(title_text = text)
    fig.show()
```

**9. TOP 10 COUNTRY REPORTS.**

```python
def topbar(Yaxis):
    country_cumulative = daily_df.groupby(
        'country'
    )[[
        'cumulative_total_cases',
        'daily_new_cases',
        'active_cases',
        'cumulative_total_deaths',
        'daily_new_deaths'
    ]].sum()
    country_cumulative.sort_values(
        by = Yaxis,
        ascending = False,
        inplace = True
    )
    country_cumulative = country_cumulative.head(10)
    fig = px.bar(
        country_cumulative,
        x = country_cumulative.index,
        y = Yaxis,
        color = country_cumulative.index,
        title = 'Top 10 Countries with Highest ' +Yaxis
    )
    return fig.update_layout(
        title_x = 0.5
    )
```

**10. COUNTRIES AND THEIR CONFIRMED CASES.**

```python
def casesbar(Yaxis):
    cases_continents = summary_df.groupby('country')[[
        'total_confirmed',
        'total_deaths',
        'total_recovered',
        'serious_or_critical',
        'total_cases_per_1m_population',
        'total_deaths_per_1m_population',
        'total_tests',
        'total_tests_per_1m_population'
    ]].mean()
    cases_continents.sort_values(
        by = Yaxis,
        ascending = False,
        inplace = True
    )
    cases_continents = cases_continents[
        cases_continents[Yaxis] > cases_continents[Yaxis].mean()
    ]
    fig = px.bar(
        cases_continents,
        x = cases_continents.index,
        y = Yaxis,
        color = cases_continents.index,
        title = 'Top Countries with Highest ' +Yaxis
    )
    return fig.update_layout(title_x = 0.5)
```

**12. COUNTRIES GRAPHS.**

```python
def country_graphs(country):
    if country in ['USA', 'UK']:
        country_prefix = 'the'
    else:
        country_prefix = ''


    country_df = daily_df[daily_df.country == country]
    country_df.set_index(
        'date',
        inplace = True
    )

    #Cumulative Total Cases Plot
    if not all(country_df.cumulative_total_cases.isna()):
        layout = go.Layout(
            yaxis = {
                'range': [
                    0,
                    country_df.cumulative_total_cases[-1] * 1.05
                ],
                'title' : 'Coronavirus Confirmed Cases'
            },
            xaxis = {
                'title' : ''
            }
        )

        fig = px.area(
            country_df,
            x = country_df.index,
            y = 'cumulative_total_cases',
            title = f"<b>Cumulative Total Confirmed Cases in {country_prefix} {country}.</b>",
            template = 'plotly_dark'
        )

        fig.update_traces(
            line = {
                'width' :5
            }
        )
        fig.update_layout(layout)
        fig.show()


    #Daily New Cases Plot
    if not all(country_df.daily_new_cases.isna()):
        layout = go.Layout(
           yaxis = {
               'range':[
                   0, country_df.daily_new_cases.max() * 1.05
               ],
                'title' : 'Daily New Coronavirus Confirmed Cases'
            },
            xaxis = {'title':''},
            template = 'plotly_dark',
            title = f"<b>Daily New Cases in  {country_prefix} {country}.</b>"
        )

        moving_avg7 = country_df.daily_new_cases.rolling(7).mean().dropna().astype(int)

        fig = go.Figure()
        fig.add_trace(
            go.Bar(
                name = 'Daily Cases',
                x = country_df.index,
                y = country_df.daily_new_cases
            )
        )

        fig.add_trace(
            go.Scatter(
                name = "7 Day Moving Average",
                x = country_df.index[country_df.shape[0] - moving_avg7.shape[0]:],
                y = moving_avg7,
                line = dict(
                    width = 3
                )
            )
        )

        fig.update_layout(layout)
        fig.show()


    #Cumulative Total Deaths Plot
    if not all(country_df.cumulative_total_deaths.isna()):
        layout = go.Layout(
            yaxis = {
                'range' : [
                    0,
                    country_df.cumulative_total_deaths[-1] * 1.05
                ],
                'title':'Coronavirus Deaths'
            },
            xaxis = {
                'title':''
            }
        )

        fig = px.area(
            country_df,
            x = country_df.index,
            y = 'cumulative_total_deaths',
            title = f"<b>Cumulative Total Deaths in {country_prefix} {country}.</b>",
            template = 'plotly_dark'
        )

        fig.update_traces(
            line = {
                'color':'red',
                'width':5
            }
        )

        fig.update_layout(layout)
        fig.show()


    #Daily New Deaths PLot
    if not all(country_df.daily_new_deaths.isna()):
        layout = go.Layout(
            yaxis = {
                'range' : [
                    0,
                    country_df.daily_new_deaths.max() * 1.05
                ],
                'title': 'Daily New Coronavirus Deaths'
            },
            xaxis = {
                'title': ''
            },
            template = 'plotly_dark',
            title = f"<b>Daily Deaths in {country_prefix} {country}.</b>"
        )

        moving_avg7 = country_df.daily_new_deaths.rolling(7).mean().dropna().astype(int)

        fig = go.Figure()

        fig.add_trace(
            go.Bar(
                name = "Daily Deaths",
                x = country_df.index,
                y = country_df.daily_new_deaths,
                marker_color = 'red'
            )
        )

        fig.add_trace(
            go.Scatter(
                name = "7 Day Moving Average",
                x = country_df.index[country_df.shape[0] - moving_avg7.shape[0]:],
                y = moving_avg7,
                line = {
                    'width' : 3,
                    'color' : 'white'
                }
            )
        )

        fig.update_layout(layout)
        fig.show()


    #Active Cases
    if not all(country_df.active_cases.isna()):
        layout = go.Layout(
            yaxis = {
                'range' :[
                    0,
                    country_df.active_cases.max() * 1.05
                ],
                'title' : 'Active Coronavirus Cases'
            },
            xaxis = {
                'title' : ''
            }
        )

        fig = px.line(
            country_df,
            x = country_df.index,
            y = "active_cases",
            title = f"<b>Active Cases in {country_prefix} {country}.</b>",
            template = 'plotly_dark'
        )

        fig.update_traces (
            line = {
                'color': 'yellow',
                'width' :5
            }
        )

        fig.update_layout(layout)
        fig.show()
```
